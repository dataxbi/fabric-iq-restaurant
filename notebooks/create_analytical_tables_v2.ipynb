```python
# Create fact_orders table
spark.sql("""
CREATE TABLE IF NOT EXISTS fact_orders (
    order_id STRING, order_date DATE, channel STRING, status STRING, delay_minutes DOUBLE,
    created_ts TIMESTAMP
) USING DELTA PARTITIONED BY (order_date)
""")
print("✓ Created fact_orders")

# Create fact_kitchen_flow table
spark.sql("""
CREATE TABLE IF NOT EXISTS fact_kitchen_flow (
    flow_id STRING, order_id STRING, station_id STRING, event_time TIMESTAMP, queue_length INT,
    created_ts TIMESTAMP
) USING DELTA PARTITIONED BY (event_time)
""")
print("✓ Created fact_kitchen_flow")

# Create dim_stations
spark.sql("""
CREATE TABLE IF NOT EXISTS dim_stations (
    station_id STRING, station_name STRING, capacity INT, created_ts TIMESTAMP
) USING DELTA
""")
print("✓ Created dim_stations")

# Populate dim_stations
spark.sql("""
INSERT INTO dim_stations (station_id, station_name, capacity, created_ts)
SELECT 'S001', 'Grill', 10, current_timestamp()
WHERE NOT EXISTS (SELECT 1 FROM dim_stations WHERE station_id = 'S001')
""")
spark.sql("""
INSERT INTO dim_stations (station_id, station_name, capacity, created_ts)
SELECT 'S002', 'Fryer', 8, current_timestamp()
WHERE NOT EXISTS (SELECT 1 FROM dim_stations WHERE station_id = 'S002')
""")
print("✓ Populated dim_stations")

print("\n✓ All tables created successfully!")
```
